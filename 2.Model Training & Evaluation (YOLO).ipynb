```python
from ultralytics import YOLO

from google.colab import drive
drive.mount('/content/drive')
```

```python
# Dataset root must contain (train, val, test) then the classes after:
DATASET_ROOT = "/content/drive/My Drive/Training_2025/Final_YOLO/source/dataset" #This folder was created to store the dataset,
                                                                                 #YOLO forces the dataset to be in the same folder under the same split neming
                                                                                 #which is (train, val, test) with no multiple test datasets, to overcome this restriction,
                                                                                 #we created this folder that we move data into when we want to train yolo or test it for the three versions.
                                                                                 #Keep in mind, each test set has to be from a single source, that is why the evaluation code is a single code, and we move data in the drive itself.

# Where YOLO will save the run
PROJECT_DIR = "/content/drive/My Drive/Training_2025/Final_YOLO/V1_MelSpect/model/" #Update it based on the experiment
RUN_NAME = "YOLO11X_V1_Final"                                                       #Update it based on the experiment
```

```python
#In case the model was already trained
model = YOLO(PROJECT_DIR+RUN_NAME+'/weights/best.pt')
```

```python
# YOLO training
model = YOLO("yolo11x-cls.pt")

model.train(
    task="classify",
    data=DATASET_ROOT,
    imgsz=224,
    model="yolo11x-cls.pt",
    batch=64,
    epochs=50,
    patience=25,
    optimizer="auto",
    lr0=1e-4,
    seed=0,
    deterministic=True,
    amp=True,
    project=PROJECT_DIR,
    name=RUN_NAME
)
```

```python
#Evaluation
metrics_test = model.val(
    data= DATASET_ROOT,
    project = PROJECT_DIR,
    name = 'Test_ASV21DF', #Update it based on the experiment
    imgsz=224,
    split='test'
)
```
